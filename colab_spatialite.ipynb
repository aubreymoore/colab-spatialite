```python
import sqlite3

# 1. Connect to an in-memory or file-based database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 2. Enable extension loading and load SpatiaLite
conn.enable_load_extension(True)
conn.load_extension('mod_spatialite.so')

# 3. Initialize spatial metadata (creates the geometry reference tables)
cursor.execute("SELECT InitSpatialMetadata(1);")

print("SpatiaLite successfully loaded!")
```

```python
# Check version
cursor.execute("SELECT spatialite_version();")
print("SpatiaLite Version:", cursor.fetchone()[0])

# Test a spatial function (creating a point geometry)
cursor.execute("SELECT AsText(MakePoint(-122.33, 47.60, 4326));")
print("Geometry output:", cursor.fetchone()[0])
```
